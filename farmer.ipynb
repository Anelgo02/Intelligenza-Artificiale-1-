## Sfida per casa

Proviamo a scrivere un agente analogo a quello delle esercitazioni per il rompicapo di capra, lupo e cavolo:

- Un contadino deve attraversare un fiume con una barca che può trasportare solo lui e (opzionalmente) uno tra una capra, un lupo o un cavolo

- Se il lupo e la capra rimangono soli  insieme, il lupo mangerà la capra

- Se la capra e il cavolo rimangono soli insieme, la capra mangerà il cavolo

Come può fare il contadino a portare tutti e tre gli oggetti sull'altra sponda del fiume ?

```python
import copy 

# Modelliamo come dizionari le condizioni iniziali

start = {
    'L' : {'f': 1, 'w' :1, 'c':1, 'g':1},
    'R' : {'f': 0, 'w' :0, 'c':0, 'g':0},
    'boat' : 'L'
}

goal = {
    'L' : {'f': 0, 'w' :0, 'c':0, 'g':0},
    'R' : {'f': 1, 'w' :1, 'c':1, 'g':1},
    'boat' : 'R'
}

# Tutte le mosse possibili, la barca deve sempre trasportare il contadino ed 
# opzionalmente un altro elemento
moves = [
    {'f': 1, 'w' :0, 'c':0, 'g':0},
    {'f': 1, 'w' :1, 'c':0, 'g':0 },
    {'f': 1, 'w' :0, 'c':1, 'g':0 },
    {'f': 1, 'w' :0, 'c':0, 'g':1 },
]



class Agent():

    def __init__(self, start=start, goal=goal, moves=moves):
        self.goal=goal
        self.state=start
        self.moves=moves
        self.frontier=[[start]]
        
    @staticmethod
    def apply_move(state, move):
        
        new_state = copy.deepcopy(state)
        new_state['boat'] = 'L' if state['boat']=='R' else 'R'

        for subject in ['f', 'w', 'c', 'g']:
            new_state[state['boat']][subject] -= move[subject]
            new_state[new_state['boat']][subject] += move[subject]
        return new_state 

    def next_states(self, state):
        def is_valid(state):
            for side in ['L', 'R']:
                if state[side]['c'] < 0 or state[side]['w'] < 0 or state[side]['f'] < 0 or state[side]['g'] < 0:
                    return False
                if state[side]['f'] == 0:    
                    if state[side]['c'] > 0 and  state[side]['g'] > 0:
                        return False
                    if state[side]['w'] > 0 and  state[side]['g'] > 0:
                        return False
            return True
        
        for move in self.moves:
            new_state = self.apply_move(state,move)
            if is_valid(new_state):
                yield new_state

    def is_goal(self,path):
        return self.goal == path[-1]

    def bfs(self):
        while self.frontier:
            path = self.frontier.pop(0)
            if self.is_goal(path):
                return path
            for s in self.next_states(path[-1]):
                if s not in path:
                    self.frontier.append(path +[s])
        
if __name__ == "__main__":
    agent = Agent()
    soluzione = agent.bfs()
    print(f"Soluzione trovata in {len(soluzione) - 1} passaggi:")
    for step, s in enumerate(soluzione):
        print(f"Passo {step}: Barca su {s['boat']} | SX: {s['L']} | DX: {s['R']}")

```
