# Problema dei tre secchi 

Sia dato il seguente problema: si hanno a disposizione due secchi di capacità rispettivamente 3 e 5 litri, inizialmente vuoti, e un grosso contenitore da 20 litri pieno d’acqua. L’obiettivo è quello di ottenere esattamente 4 litri nel secchio da 5 litri.

Le azioni possibili sono:
- riempire un secchio dal contenitore;
- svuotare un secchio nel contenitore;
- travasare da un secchio all’altro finché uno non è pieno o l’altro non è vuoto.

```python
class Agent():
    def __init__(self, jugs=(3,5,20), start = (0,0,20), target=4):
        self.jugs = jugs
        self.target = target
        self.frontier = [[start]]

    # La nostra azione è univocamente determinata da sorgente e destinazione, 
    # quindi possiamo rappresentarla come una tupla (src, dst).

    def next_states(self,state):
        def is_valid(state):
            return all(0 <= curr <= cap for curr,cap in zip(state, self.jugs))
        
        # moves genera tutte le possibili coppie sorgente-destinazione tra i secchi.
        moves = ((i,j) for i in range(len(self.jugs)) for j 
                 in range(len(self.jugs)) if i != j)
        
        # La condizione i != j esclude travasi da un secchio verso sé stesso.

        for src,dst in moves:
            amount = min(state[src], self.jugs[dst] - state[dst])
            new_state = list(state)
            new_state[src] -= amount
            new_state[dst] += amount
            if is_valid(new_state):
                yield tuple(new_state) 

    def is_goal(self,path):
        return path[-1][1] == self.target

    def bfs(self):
        while self.frontier:
            path = self.frontier.pop(0)
            if self.is_goal(path):
                return path
            for s in self.next_states(path[-1]):
                if s not in path:
                    self.frontier.append(path + [s])

if __name__ == "__main__":
    agent = Agent()
    print(agent.bfs())
```

self.frontier rappresenta la **frontiera** della ricerca: *i percorsi che l’algoritmo deve ancora esplorare*.

C’è un solo percorso da esplorare, composto solo dallo stato iniziale. Quando l’algoritmo trova nuovi stati, può aggiungerli al percorso e mettere quel percorso nella frontiera. Così conserva anche la strada seguita, utile per ricostruire la soluzione.

Con una sola lista, self.frontier = [start], la frontiera conterrebbe invece solo stati singoli, non percorsi. Questa forma è possibile, ma per ricostruire il cammino servirebbe memorizzare le relazioni tra stati in un altro modo.
